```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
daily_sales = np.array([250.50, 300.20, 180.00, 450.75, 520.10, 299.99, 375.60])

average_sales = daily_sales.mean()
max_sale = daily_sales.max()
min_sale = daily_sales.min()
number_of_days = daily_sales.size

print(f"Sales Data: {daily_sales}")
print(f"Number of Days: {number_of_days}")
print(f"Average Daily Sales: ${average_sales:.2f}")
print(f"Max Single-Day Sale: ${max_sale:.2f}")
print(f"Min Single-Day Sale: ${min_sale:.2f}")
```

```python
student_data = {
    'StudentID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [20, 21, 19, 22, 20],
    'Major': ['CompSci', 'Math', 'Physics', 'Chemistry', 'Biology']
}

student_roster_df = pd.DataFrame(student_data)

print("Student Roster Head:")
print(student_roster_df.head())

print(f"\nRoster Dimensions (rows, columns): {student_roster_df.shape}")
```

```python
months = ['January', 'February', 'March', 'April', 'May']
profits = [20000, 25000, 22000, 30000, 28000]

plt.bar(months, profits, color='skyblue')

plt.title('Monthly Profits for First 5 Months')
plt.xlabel('Month')
plt.ylabel('Profit ($)')
plt.show()
```

```python
temp_grid = np.array([
    [25, 28, 30, 26],
    [24, 27, 29, 25],
    [26, 29, 31, 27],
    [23, 26, 28, 24]
])
print("Original Temperature Grid (4x4):")
print(temp_grid)

inner_core = temp_grid[1:3, 1:3]
print("\nSliced Inner 2x2 Core:")
print(inner_core)

reshaped_core = inner_core.reshape(4)
print("\nReshaped Core (1x4):")
print(reshaped_core)

# Slice to get the first column
first_column = temp_grid[:, 0]
print("\nFirst Column:")
print(first_column)
```

```python
inventory_data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam'],
    'Category': ['Electronics', 'Accessory', 'Accessory', 'Electronics', 'Accessory'],
    'Price': [1200, 25, 75, 300, 45],
    'Stock': [8, 150, 75, 20, 9]
}
inventory_df = pd.DataFrame(inventory_data)

print("Original Inventory DataFrame:")
print(inventory_df)

low_stock_electronics = inventory_df[(inventory_df['Stock'] < 10) & (inventory_df['Category'] == 'Electronics')]

print("\nFiltered Low-Stock Electronics:")
print(low_stock_electronics)
```
